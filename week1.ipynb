```python
import numpy as np
class NeuralNetwork:
  def __init__(self, i_size, h_size, o_size):
    # Concise parameter initialization using tuple unpacking
    self.w1, self.b1 = np.random.randn(i_size, h_size), np.ones((1, h_size))
    self.w2, self.b2 = np.random.randn(h_size, o_size), np.ones((1, o_size))
    print(self.b1)
    print(self.b2)
  def forward(self,x):
    sig = lambda x:1 /(1+np.exp(-x))
    return sig(sig(x @ self.w1 + self.b1) @ self.w2 + self.b2)
```

```python
# --- Example Usage ---
if __name__ == "__main__":
  # 1. Define network architecture (3 Inputs, 5 Hidden units, 2 Outputs)
  nn = NeuralNetwork(i_size=3, h_size=5, o_size=2)
  # 2. Replicate your precise sample input (Batch size of 2, with 3 features each)
  x_input = np.array([[0.1, 0.5, 0.9],
                      [0.2, 0.8, 0.6]])
```

```python
# 3. Process the matrix data through the network
output = nn.forward(x_input)

# 4. Show the prediction matrices
print("Output shape:", output.shape)

print("Network predictions:\n", output)
```
