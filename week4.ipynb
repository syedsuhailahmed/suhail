```python
import numpy as np
#1 / 1 + e^-x = 0 - 1
def lonsig(z):
    return 1 / 1+np.exp(-z)

# Input vector
X = np.array([0.7, -1.3, 3.5])

# Weight vector
W = np.array([0.8, -0.5, 2.0])

# Bias
b = 0.2

# Calculate weighted sum
Z = np.dot(W, X) + b

# Apply lonsig activation function
output_lonsig = lonsig(Z)

# Display results
print("Weighted sum (Z):", Z)
print("Output with lonsig:", output_lonsig)
```
