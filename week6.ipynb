```python
import numpy as np
import matplotlib.pyplot as plt

# XOR Gate Dataset - Non-linearly separable
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
y = np.array([0, 1, 1, 0])
```

```python
# Perceptron Class
class Perceptron:
    def __init__(self, lr=0.1, epochs=10):
        self.lr = lr
        self.epochs = epochs
        self.w = np.zeros(X.shape[1])
        self.b = 0

    def activation(self, z):
        return np.where(z >= 0, 1, 0)

    def fit(self, X, y):
        print("Training...\n")

        for epoch in range(self.epochs):
            errors = 0

            for xi, target in zip(X, y):

                # Calculate prediction
                z = np.dot(xi, self.w) + self.b
                pred = self.activation(z)

                # Calculate error
                error = target - pred
                # Update weights and bias
                if error != 0:
                    self.w += self.lr * error * xi
                    self.b += self.lr * error
                    errors += 1

            print( f"Epoch {epoch + 1}:,Errors = {errors}, Weights = {self.w}, Bias = {self.b}" )

        print("\nTraining stopped.")
        print("For XOR, the perceptron cannot achieve zero errors.")

    def predict(self, X):
        return self.activation(
            np.dot(X, self.w) + self.b
        )
```

```python
# Create and train perceptron
p = Perceptron(lr=0.1, epochs=10)
p.fit(X, y)

# Final predictions
predictions = p.predict(X)

print("\nActual Output:    ", y)
print("Predicted Output: ", predictions)
```
