```python
import numpy as np

# Tansig (Hyperbolic Tangent Sigmoid) Activation Function
def tansig(z):
    return 2 / (1+np.exp(-2*z))-1
    # Equivalent to: np.tanh(z)

# Input vector
X = np.array([0.7, -1.3, 3.5])

# Weight vector
W = np.array([0.8, -0.5, 2.0])

# Bias
b = 0.2

# Calculate weighted sum
Z = np.dot(W, X) + b

# Apply Tansig activation function
output_tansig = tansig(Z)

# Display results
print("Weighted sum (Z):", Z)
print("Output with Tansig:", output_tansig)
```
