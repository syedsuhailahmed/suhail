```python
import numpy as np

# Input data
x = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]]) #[4,2]

# Labels: AND gate
y = np.array([0, 0, 0, 1])
```

```python
class Perceptron:

    def __init__(self, learning_rate=0.1, epochs=10):
        self.lr = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = None

    # Step activation function
    #if z=pov=1 ; if z = neg= 0
    def _activation(self, z):
        return np.where(z >= 0, 1, 0)

    # Training
    def fit(self, x, y):

        # Create one weight for each input feature
        self.weights = np.zeros(x.shape[1]) #[0,0]
        self.bias = 0.0

        print("\nStarting training:")

        for epoch in range(self.epochs):

            errors_in_epoch = 0

            for idx, x_i in enumerate(x): #[1,2,3,4]

                # Weighted sum
                linear_output = np.dot(x_i, self.weights) + self.bias

                # Prediction
                prediction = self._activation(linear_output)

                # Calculate error
                error = y[idx] - prediction

                # Update weights and bias
                if error != 0:
                    self.weights += self.lr * error * x_i
                    self.bias += self.lr * error
                    errors_in_epoch += 1

            print(f"Epoch: {epoch + 1}/{self.epochs}-classification Errors: {errors_in_epoch}")

            # Stop if no errors
            if errors_in_epoch == 0:
                print(f"--> Converged successfully at epoch {epoch + 1}!")
                break

    # Prediction
    def predict(self, x):
        linear_output = np.dot(x, self.weights) + self.bias
        return self._activation(linear_output)


# Create perceptron
clf = Perceptron(learning_rate=0.1, epochs=10)

# Train
clf.fit(x, y)

# Test
predictions = clf.predict(x)

print("\nPredictions:", predictions)
print("Actual:", y)
```
